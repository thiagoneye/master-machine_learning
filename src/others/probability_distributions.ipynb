## Imports

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

## Discrete Probability Distributions

In this section, we visualize distributions for discrete random variables.

We use the Probability Mass Function (PMF), which gives the probability of a specific outcome.

Bar plots are used for visualization.

```python
plt.style.use("seaborn-v0_8-whitegrid")

# Create a figure with a 1x4 grid of subplots
fig, axs = plt.subplots(1, 4, figsize=(20, 5))
fig.suptitle("Common Probability Distributions", fontsize=20)

# Bernoulli Distribution
p_bernoulli = 0.7
x_bernoulli = np.array([0, 1])
y_bernoulli = stats.bernoulli.pmf(x_bernoulli, p=p_bernoulli)
axs[0].bar(x_bernoulli, y_bernoulli, color="skyblue")
axs[0].set_title(f"Bernoulli Distribution (p={p_bernoulli})", fontsize=12)
axs[0].set_xlabel("Outcome (0=Failure, 1=Success)")
axs[0].set_ylabel("Probability")
axs[0].set_xticks([0, 1])
axs[0].grid(False)
axs[0].spines["top"].set_visible(False)
axs[0].spines["right"].set_visible(False)

# Binomial Distribution
n_binomial, p_binomial = 20, 0.5
x_binomial = np.arange(0, n_binomial + 1)
y_binomial = stats.binom.pmf(x_binomial, n=n_binomial, p=p_binomial)
axs[1].bar(x_binomial, y_binomial, color="salmon")
axs[1].set_title(
    f"Binomial Distribution (n={n_binomial}, p={p_binomial})", fontsize=12
)
axs[1].set_xlabel("Number of Successes")
axs[1].set_ylabel("Probability")
axs[1].grid(False)
axs[1].spines["top"].set_visible(False)
axs[1].spines["right"].set_visible(False)

# Poisson Distribution
mu_poisson = 4
x_poisson = np.arange(0, 15)
y_poisson = stats.poisson.pmf(x_poisson, mu=mu_poisson)
axs[2].bar(x_poisson, y_poisson, color="lightgreen")
axs[2].set_title(f"Poisson Distribution (μ={mu_poisson})", fontsize=12)
axs[2].set_xlabel("Number of Events in Interval")
axs[2].set_ylabel("Probability")
axs[2].grid(False)
axs[2].spines["top"].set_visible(False)
axs[2].spines["right"].set_visible(False)

# Geometric Distribution
p_geometric = 0.3
x_geometric = np.arange(1, 15)
y_geometric = stats.geom.pmf(x_geometric, p=p_geometric)
axs[3].bar(x_geometric, y_geometric, color="plum")
axs[3].set_title(f"Geometric Distribution (p={p_geometric})", fontsize=12)
axs[3].set_xlabel("Trials until First Success")
axs[3].set_ylabel("Probability")
axs[3].grid(False)
axs[3].spines["top"].set_visible(False)
axs[3].spines["right"].set_visible(False)

# Final Adjustments & Display
# Adjust layout to prevent labels from overlapping
plt.tight_layout(rect=[0, 0, 1, 0.96])
# Show the final plot
plt.show()
```

## Continuous Probability Distributions

In this section, we visualize distributions for continuous random variables.

We use the Probability Density Function (PDF), where the area under the curve represents probability.

Line plots are used for visualization.

```python
plt.style.use("seaborn-v0_8-whitegrid")

# Create a figure with a 1x4 grid of subplots
fig, axs = plt.subplots(1, 4, figsize=(20, 5))
fig.suptitle("Common Probability Distributions", fontsize=20)

# Normal Distribution
mu_normal, sigma_normal = 0, 1
x_normal = np.linspace(-4, 4, 500)
y_normal = stats.norm.pdf(x_normal, loc=mu_normal, scale=sigma_normal)
axs[0].plot(x_normal, y_normal, color="dodgerblue")
axs[0].set_title(
    f"Normal Distribution (μ={mu_normal}, σ={sigma_normal})", fontsize=12
)
axs[0].set_xlabel("Value")
axs[0].set_ylabel("Density")
axs[0].grid(False)
axs[0].spines["top"].set_visible(False)
axs[0].spines["right"].set_visible(False)

# Uniform Distribution
a_uniform, b_uniform = -2, 5
x_uniform = np.linspace(-3, 6, 500)
y_uniform = stats.uniform.pdf(x_uniform, loc=a_uniform, scale=(b_uniform - a_uniform))
axs[1].plot(x_uniform, y_uniform, color="orangered")
axs[1].set_title(f"Uniform Distribution (a={a_uniform}, b={b_uniform})", fontsize=12)
axs[1].set_xlabel("Value")
axs[1].set_ylabel("Density")
axs[1].grid(False)
axs[1].spines["top"].set_visible(False)
axs[1].spines["right"].set_visible(False)

# Exponential Distribution
lambda_exp = 0.5  # Rate parameter
scale_exp = 1 / lambda_exp  # Scale parameter for scipy
x_exp = np.linspace(0, 10, 500)
y_exp = stats.expon.pdf(x_exp, scale=scale_exp)
axs[2].plot(x_exp, y_exp, color="green")
axs[2].set_title(f"Exponential Distribution (λ={lambda_exp})", fontsize=12)
axs[2].set_xlabel("Value")
axs[2].set_ylabel("Density")
axs[2].grid(False)
axs[2].spines["top"].set_visible(False)
axs[2].spines["right"].set_visible(False)

# Chi-Squared Distribution
x_chi2 = np.linspace(0, 20, 500)
dfs = [2, 4, 8]  # Degrees of freedom
colors = ["purple", "gold", "brown"]
for df, color in zip(dfs, colors):
    y_chi2 = stats.chi2.pdf(x_chi2, df=df)
    axs[3].plot(x_chi2, y_chi2, label=f"df={df}", color=color)

axs[3].set_title("Chi-Squared Distribution", fontsize=12)
axs[3].set_xlabel("Value")
axs[3].set_ylabel("Density")
axs[3].legend()
axs[3].set_ylim(0, 0.5)  # Set y-limit for better visualization
axs[3].grid(False)
axs[3].spines["top"].set_visible(False)
axs[3].spines["right"].set_visible(False)
axs[3].legend(frameon=True)

# Final Adjustments & Display
# Adjust layout to prevent labels from overlapping
plt.tight_layout(rect=[0, 0, 1, 0.96])
# Show the final plot
plt.show()
```

## Box-Cox Transformation

```python
plt.style.use("seaborn-v0_8-whitegrid")

np.random.seed(42)
original_data = stats.chi2.rvs(df=4, size=500) + 1

# Apply the Box-Cox Transformation

# The stats.boxcox function finds the optimal lambda and performs the transformation.
# It returns the transformed data and the optimal lambda value.
transformed_data, optimal_lambda = stats.boxcox(original_data)

# Print the optimal lambda found by the function.
print(f"Optimal lambda found: {optimal_lambda:.2f}")

# Plots

# We'll use histograms and Q-Q plots to compare the distributions.
# A Q-Q plot compares the quantiles of the data to the quantiles of a normal distribution.
# If the data is normal, the points should fall closely along the red line.

# Setup the plot figure
fig, axes = plt.subplots(2, 2, figsize=(8, 6))
fig.suptitle("Box-Cox Transformation Example", fontsize=16)

# Before Transformation

# Histogram of the original data
axes[0, 0].hist(
    original_data, bins=30, color="skyblue", edgecolor="black", density=True
)
axes[0, 0].set_title("Histogram of Original Data (Skewed)")
axes[0, 0].set_xlabel("Value")
axes[0, 0].set_ylabel("Frequency")
axes[0, 0].grid(False)
axes[0, 0].spines["top"].set_visible(False)
axes[0, 0].spines["right"].set_visible(False)

# Q-Q plot of the original data
stats.probplot(original_data, dist="norm", plot=axes[0, 1])
axes[0, 1].set_title("Q-Q Plot of Original Data")
axes[0, 1].grid(False)
axes[0, 1].spines["top"].set_visible(False)
axes[0, 1].spines["right"].set_visible(False)

# After Transformation

# Histogram of the transformed data
axes[1, 0].hist(
    transformed_data, bins=30, color="lightgreen", edgecolor="black", density=True
)
axes[1, 0].set_title("Histogram of Transformed Data (Box-Cox)")
axes[1, 0].set_xlabel("Transformed Value")
axes[1, 0].set_ylabel("Frequency")
axes[1, 0].grid(False)
axes[1, 0].spines["top"].set_visible(False)
axes[1, 0].spines["right"].set_visible(False)

# Q-Q plot of the transformed data
stats.probplot(transformed_data, dist="norm", plot=axes[1, 1])
axes[1, 1].set_title("Q-Q Plot of Transformed Data")
axes[1, 1].grid(False)
axes[1, 1].spines["top"].set_visible(False)
axes[1, 1].spines["right"].set_visible(False)

# Display the plots
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
